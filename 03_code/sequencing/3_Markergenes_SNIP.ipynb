```r
suppressPackageStartupMessages({
library(Seurat)
library(SeuratWrappers)
library(patchwork)
library(dplyr)
library(ggplot2)
})

theme_set(theme_classic(16))

# Set working directory for saving and accessing files
setwd("/data/pam/team333/skb/scratch/SNIP_paper/04_analysis/")
set.seed(123)
```

```r
Hcon_obj <- readRDS(file = "seurat_obj/Hcon_harmony.rds")
Tmur_obj <- readRDS(file = "seurat_obj/Tmur_harmony.rds")
```

```r
Hcon_join <- JoinLayers(Hcon_obj)
Tmur_join <- JoinLayers(Tmur_obj)
```

```r
# Find marker genes for each cluster of Seurat objects at res = 0.8
Hcon_markers <- FindAllMarkers(Hcon_join, only.pos = TRUE)
Tmur_markers <- FindAllMarkers(Tmur_join, only.pos = TRUE)
```

```r
Hcon_markers_filt <- Hcon_markers %>%
    group_by(cluster) %>%
    dplyr::filter(avg_log2FC > 1 & p_val_adj < 0.05)
write.csv(Hcon_markers_filt, "Hcon_filt_all.markers.txt")

Tmur_markers_filt <- Tmur_markers %>%
    group_by(cluster) %>%
    dplyr::filter(avg_log2FC > 1 & p_val_adj < 0.05)
write.csv(Tmur_markers_filt, "Tmur_filt_all.markers.txt")
```
